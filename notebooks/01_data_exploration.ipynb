```python

from pathlib import Path
import pandas as pd

# Notebook'un çalıştığı klasörü gör
print("Çalışma klasörü:", Path.cwd())

# notebooks klasöründen bir üst dizine çık
BASE_DIR = Path.cwd().parent

IMAGE_DIR = BASE_DIR / "images"
MASK_DIR = BASE_DIR / "masks"
CSV_PATH = BASE_DIR / "GroundTruth.csv"

print("CSV bulundu mu?", CSV_PATH.is_file())
print("Images bulundu mu?", IMAGE_DIR.is_dir())
print("Masks bulundu mu?", MASK_DIR.is_dir())

# CSV'yi oku
df = pd.read_csv(CSV_PATH)

# İlk 5 satırı göster
display(df.head())

# Veri setinin boyutu
print("Veri seti boyutu:", df.shape)
len(df)
```

```python
print("Tekrarlanan ID:", df["image"].duplicated().sum())
```

```python
classes = ["MEL", "NV", "BCC", "AKIEC", "BKL", "DF", "VASC"]
df["label"] = df[classes].idxmax(axis=1)
df["label"]
```

```python
# Her satırda tam olarak bir sınıf işaretlenmiş mi?
class_count = df[classes].sum(axis=1)

print("Geçersiz sınıf etiketi sayısı:",
      (class_count != 1).sum())
```

```python
df["image_path"] = df["image"].apply(
    lambda image_id: IMAGE_DIR / f"{image_id}.jpg"
)

display(df[["image", "label", "image_path"]].head())
```

```python
image_exists = df["image_path"].map(lambda p: p.is_file())

print("Eksik görüntü:", (~image_exists).sum())
```

```python
mask_lookup = {
    p.stem.replace("_segmentation", ""): p
    for p in MASK_DIR.iterdir()
    if p.is_file() and "_segmentation" in p.stem
}

print("Bulunan maske sayısı:", len(mask_lookup))
```

```python
#Maskeleri DataFrame'e ekleme#
df["mask_path"] = df["image"].map(mask_lookup)

display(
    df[["image", "label", "image_path", "mask_path"]].head()
)
```

```python


# Maske eşleşmesi var mı?
print("Eksik maske sayısı:",
      df["mask_path"].isna().sum())


```
